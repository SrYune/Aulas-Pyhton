## 1) Exercício sobre strings

```python
def removerCaractere(texto, indice):
    """Remove o caractere na posição `indice` de `texto` e retorna a string com o resultado."""
    # Fatia 1: tudo antes do índice. Fatia 2: tudo depois do índice.
    return texto[:indice] + texto[indice + 1:]
```

```python
print(removerCaractere('casa', 2))
print(removerCaractere('casa', 0))
print(removerCaractere('casa', len('casa') - 1))
```

## 2) Exercício sobre operadores aritméticos

```python
def volumeDaEsfera(raio):
    """Retorna o volume de uma esfera de raio `raio`, usando pi = 3.14: V = (4/3) * pi * r^3."""
    PI = 3.14
    return (4 / 3) * pi * raio ** 3
```

## 3) Exercício sobre laços de repetição

```python
def exponenciação(base, expoente):
    """Calcula base elevada a expoente (expoente inteiro >= 0)."""
    resultado = 1
    for _ in range(expoente):
        resultado *= base  # multiplica a base 'expoente' vezes
    return resultado
```

```python
print('Retorno:', exponenciação(2, 0), ' - Valor correto?', exponenciação(2, 0)==1)
print('Retorno:', exponenciação(2, 1), ' - Valor correto?', exponenciação(2, 1)==2)
print('Retorno:', exponenciação(2, 2), ' - Valor correto?', exponenciação(2, 2)==4)
print('Retorno:', exponenciação(2, 3), ' - Valor correto?', exponenciação(2, 3)==8)
print('Retorno:', exponenciação(2, 4), ' - Valor correto?', exponenciação(2, 4)==16)
```

## 4) Exercício sobre laços de repetição (Fibonacci)

```python
# Gera e imprime a série de Fibonacci até que o valor gerado seja maior que 100000.
anterior, atual = 0, 1  # F0 e F1
serie = [anterior]
while atual <= 100000:
    serie.append(atual)
    anterior, atual = atual, anterior + atual  # F(n) = F(n-1) + F(n-2)

print(serie)
```

## 5) Exercício sobre controle de fluxo e operadores de associação

```python
def éVogal(letra):
    """Retorna True se `letra` for uma vogal (maiúscula ou minúscula) e False caso contrário."""
    vogais = ['a', 'e', 'i', 'o', 'u']
    return letra.lower() in vogais
```

```python
print('Retorno:', éVogal('v'), ' - Valor correto?', éVogal('v')==False)
print('Retorno:', éVogal('A'), ' - Valor correto?', éVogal('A')==True)
```

## 6) Exercício sobre controle de fluxo, operadores lógicos e de comparação

```python
def calcularMédia(nota1, nota2):
    """Calcula a média de duas notas e imprime as notas, a média, o conceito e o resultado (APROVADO/REPROVADO)."""
    media = (nota1 + nota2) / 2

    # Atribuição do conceito conforme a tabela 
    if 9.0 <= media <= 10.0:
        conceito = 'A'
    elif 7.5 <= media < 9.0:
        conceito = 'B'
    elif 6.0 <= media < 7.5:
        conceito = 'C'
    elif 4.0 <= media < 6.0:
        conceito = 'D'
    else:
        conceito = 'E'

    # A, B e C aprovam; D e E reprovam
    resultado = 'APROVADO' if conceito in ['A', 'B', 'C'] else 'REPROVADO'

    print('Nota 1: %.2f - Nota 2: %.2f - Média: %.2f - Conceito: %s - Resultado: %s' % (nota1, nota2, media, conceito, resultado))
```

```python
calcularMédia(1, 1)
calcularMédia(5, 5)
calcularMédia(6, 6)
calcularMédia(8, 8)
calcularMédia(10, 10)
```

## 7) Exercício sobre funções (IMC com pares chave-valor)

```python
def imc(peso, altura):
    """Calcula o IMC = peso / (altura * altura). Os parâmetros podem ser passados por chave (peso=..., altura=...)."""
    return peso / (altura * altura)
```

```python
print('Resultado:', imc(peso=100, altura=2.0), ' - Teste está correto?', imc(peso=100, altura=2.0)==25.0)
print('Resultado:', imc(altura=2.0, peso=100), ' - Teste está correto?', imc(altura=2.0, peso=100)==25.0)
```

## 8) Exercício sobre funções (número arbitrário de parâmetros)

> Obs.: esta função tem o mesmo nome da do exercício 6 e, ao ser executada, passa a substituir aquela. O exercício 6 já foi testado antes, então não há problema.

```python
def calcularMédia(*valores):
    """Retorna a média aritmética simples de qualquer quantidade de valores recebidos."""
    return sum(valores) / len(valores)
```

```python
print('Resultado:', calcularMédia(2), ' - Teste está correto?', calcularMédia(2)==2)
print('Resultado:', calcularMédia(2, 4), ' - Teste está correto?', calcularMédia(2, 4)==3)
print('Resultado:', calcularMédia(2, 4, 6), ' - Teste está correto?', calcularMédia(2, 4, 6)==4)
print('Resultado:', calcularMédia(1, 2, 3, 4), ' - Teste está correto?', calcularMédia(1, 2, 3, 4)==2.5)
```

## 9) Exercício sobre funções (fatorial recursivo)

```python
def fatorial(x):
    """Calcula o fatorial de x usando recursão."""
    if x <= 1:  # caso base: 0! = 1 e 1! = 1
        return 1
    return x * fatorial(x - 1)
```

```python
print('fatorial:', fatorial(0), ' - resultado correto?', fatorial(0)==1)
print('fatorial:', fatorial(1), ' - resultado correto?', fatorial(1)==1)
print('fatorial:', fatorial(2), ' - resultado correto?', fatorial(2)==2)
print('fatorial:', fatorial(3), ' - resultado correto?', fatorial(3)==6)
print('fatorial:', fatorial(4), ' - resultado correto?', fatorial(4)==24)
```

## 10) Exercício sobre funções (soma de valores)

```python
def somarValores(*valores):
    """Retorna a soma de qualquer quantidade de valores recebidos."""
    soma = 0
    for valor in valores:
        soma += valor
    return soma
```

```python
print('Resultado:', somarValores(2), ' - Teste está correto?', somarValores(2)==2)
print('Resultado:', somarValores(2, 4), ' - Teste está correto?', somarValores(2, 4)==6)
print('Resultado:', somarValores(2, 4, 6), ' - Teste está correto?', somarValores(2, 4, 6)==12)
print('Resultado:', somarValores(1, 2, 3, 4), ' - Teste está correto?', somarValores(1, 2, 3, 4)==10)
print('Resultado:', somarValores(1, 2, 3, 4, 5), ' - Teste está correto?', somarValores(1, 2, 3, 4, 5)==15)
```
